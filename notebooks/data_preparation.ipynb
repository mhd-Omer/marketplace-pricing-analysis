# Data Preparation Notebook

Purpose:
- Prepare raw Inside Airbnb CSV files for PostgreSQL ingestion
- Perform minimal schema alignment and type coercion
- No analytical logic is performed here

All analysis is conducted in SQL.

```python
import pandas as pd
```

```python
df = pd.read_csv('../data/listings_nyc.csv')
df.shape
```

```python
columns = [
    "id",
    "host_id",
    "host_is_superhost",
    "room_type",
    "property_type",
    "accommodates",
    "price",
    "minimum_nights",
    "availability_365",
    "number_of_reviews",
    "reviews_per_month",
    "neighbourhood",
    "neighbourhood_cleansed",
    "latitude",
    "longitude"
]

df = df[columns]
df.head()
```

```python
# Remove $ and commas from price
df["price"] = (
    df["price"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

# Convert numeric columns safely
numeric_cols = [
    "price",
    "accommodates",
    "minimum_nights",
    "availability_365",
    "number_of_reviews",
    "reviews_per_month",
    "latitude",
    "longitude"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df.info()
```

```python
df.to_csv("../data/listings_nyc_clean.csv", index=False)
```
